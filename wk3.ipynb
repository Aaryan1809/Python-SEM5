```python
num = float(input("Enter a number: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

```python
n=int(input(num))
sum=0
count=1

while(count<=n):
    sum=sum+count
    count=count+1

print(sum)
```
